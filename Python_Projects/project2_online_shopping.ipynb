```python

# Online Shopping Checkout

name = input("Enter customer name: ")
amount = float(input("Enter purchase amount: "))
customer_type = input("Enter customer type (Regular/Premium): ").strip().lower()

if amount < 0:
    print("Invalid purchase amount!")
elif customer_type not in ("regular", "premium"):
    print("Invalid customer type!")
else:
    if amount >= 5000:
        discount_rate = 20
    elif amount >= 3000:
        discount_rate = 10
    elif amount >= 1000:
        discount_rate = 5
    else:
        discount_rate = 0

    discount = amount * discount_rate / 100
    amount_after_discount = amount - discount

    if customer_type == "premium":
        extra_discount = amount_after_discount * 5 / 100
    else:
        extra_discount = 0

    total_discount = discount + extra_discount
    amount_after_discount -= extra_discount

    if amount_after_discount >= 2000:
        delivery_charge = 0
    else:
        delivery_charge = 100

    final_amount = amount_after_discount + delivery_charge

    print("\n--- Shopping Bill ---")
    print("Customer Name:", name)
    print("Customer Type:", customer_type.title())
    print("Discount Amount: Rs.", round(total_discount, 2))
    print("Delivery Charge: Rs.", delivery_charge)
    print("Final Payable Amount: Rs.", round(final_amount, 2))
```
