```python

# Student Scholarship Eligibility

name = input("Enter student name: ")
percentage = float(input("Enter percentage: "))
income = float(input("Enter annual family income: "))
attendance = float(input("Enter attendance percentage: "))

if not (0 <= percentage <= 100 and 0 <= attendance <= 100 and income >= 0):
    print("Invalid input! Check percentage, attendance or income.")
else:
    if percentage >= 75 and income <= 400000 and attendance >= 75:
        if percentage >= 90:
            category = "Full Scholarship"
        else:
            category = "Partial Scholarship"

        print("\nStudent Name:", name)
        print("Eligibility Status: Eligible")
        print("Scholarship Category:", category)
    else:
        print("\nStudent Name:", name)
        print("Eligibility Status: Not Eligible")

        if percentage < 75:
            print("Reason: Percentage is below 75%.")
        elif income > 400000:
            print("Reason: Family income exceeds Rs. 400000.")
        elif attendance < 75:
            print("Reason: Attendance is below 75%.")
```
