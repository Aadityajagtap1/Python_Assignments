```python

# Restaurant Bill and Service Charge

name = input("Enter customer name: ")
food_bill = float(input("Enter food bill amount: "))
membership = input("Are you a member? (Yes/No): ").strip().lower()

if food_bill < 0:
    print("Invalid bill amount!")
elif membership not in ("yes", "no"):
    print("Invalid membership status!")
else:
    if food_bill >= 5000:
        discount_rate = 15
    elif food_bill >= 3000:
        discount_rate = 10
    elif food_bill >= 1000:
        discount_rate = 5
    else:
        discount_rate = 0

    discount = food_bill * discount_rate / 100
    amount_after_discount = food_bill - discount

    if membership == "yes":
        member_discount = amount_after_discount * 5 / 100
    else:
        member_discount = 0

    total_discount = discount + member_discount
    discounted_bill = food_bill - total_discount

    service_charge = discounted_bill * 5 / 100
    final_amount = discounted_bill + service_charge

    print("\n--- Restaurant Bill ---")
    print("Customer Name:", name)
    print("Original Bill: Rs.", food_bill)
    print("Total Discount: Rs.", round(total_discount, 2))
    print("Service Charge: Rs.", round(service_charge, 2))
    print("Final Payable Amount: Rs.", round(final_amount, 2))

    if final_amount >= 5000:
        print("Thank you for being a valued customer")
    else:
        print("Thank you for visiting!")
```
