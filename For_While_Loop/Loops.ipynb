```python

# Q13. Check whether a number is palindrome.
n = int(input("Enter number: "))
temp = n
rev = 0
while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10
if temp == rev:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
# Q28. Check whether a number is Armstrong.
n = int(input("Enter number: "))
temp = n
total = 0
digits = len(str(n))
while n > 0:
    digit = n % 10
    total = total + digit ** digits
    n = n // 10
if temp == total:
    print("Armstrong number")
else:
    print("Not Armstrong number")
```

```python
# Q47. Find product of digits of a number.
n = abs(int(input("Enter number: ")))
product = 1
if n == 0:
    product = 0
while n > 0:
    digit = n % 10
    product = product * digit
    n = n // 10
print(product)
```

```python
# Q48. Check whether a number is Strong number.
n = int(input("Enter number: "))
temp = n
total = 0
while n > 0:
    digit = n % 10
    fact = 1
    for i in range(1, digit + 1):
        fact = fact * i
    total = total + fact
    n = n // 10
if temp == total:
    print("Strong number")
else:
    print("Not strong number")
```

```python
# Q49. Display menu-driven program using while loop.
while True:
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Exit")

    choice = int(input("Enter choice: "))

    if choice == 4:
        break

    if choice < 1 or choice > 4:
        print("Invalid choice")
        continue

    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if choice == 1:
        print(a + b)
    elif choice == 2:
        print(a - b)
    elif choice == 3:
        print(a * b)
```
